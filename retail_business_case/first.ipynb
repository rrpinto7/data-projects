```python
import pandas as pd
import numpy as np
```

```python
# 1. Carregar dataset (Excel no teu caso)
file_path = "Data_labels (1).xlsx"
df = pd.read_excel(file_path)
```

```python
# 2. Info geral
print("### DIMENSÕES ###")
print(f"Linhas: {df.shape[0]}, Colunas: {df.shape[1]}\n")

print("### TIPOS DE VARIÁVEIS ###")
print(df.dtypes, "\n")

# 3. Estatísticas descritivas
print("### ESTATÍSTICAS DESCRITIVAS (numéricas) ###")
print(df.describe().T, "\n")

print("### ESTATÍSTICAS DESCRITIVAS (categóricas) ###")
print(df.describe(include=['object', 'category']).T, "\n")

# 4. Valores em falta
print("### VALORES NULOS ###")
nulls = df.isnull().sum()
print(nulls[nulls > 0].sort_values(ascending=False), "\n")

# 5. Duplicados
print("### DUPLICADOS ###")
print(f"Nº de linhas duplicadas: {df.duplicated().sum()}\n")

# 6. Valores únicos por coluna
print("### VALORES ÚNICOS ###")
for col in df.columns:
    print(f"{col}: {df[col].nunique()} valores únicos")
```
